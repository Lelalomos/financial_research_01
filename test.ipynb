```python
import yfinance as yf
import pandas as pd
import numpy as np

# Download stock data
df = yf.download(
    "AAPL",
    start="2022-01-01",
    end="2024-01-01",
    progress=False,
    multi_level_index= False
)

df = df.reset_index()
```

```python
df
```

```python
def add_fibonacci_features(df, window=60):
    df = df.copy()

    # Rolling swing high / low
    df["swing_high"] = df["High"].rolling(window).max()
    df["swing_low"] = df["Low"].rolling(window).min()

    # Price range
    df["range"] = df["swing_high"] - df["swing_low"]

    # Fibonacci retracement levels
    df["fib_38"] = df["swing_high"] - 0.382 * df["range"]
    df["fib_50"] = df["swing_high"] - 0.5 * df["range"]
    df["fib_61"] = df["swing_high"] - 0.618 * df["range"]

    # Normalized distance features (RNN-friendly)
    df["dist_fib_38"] = (df["Close"] - df["fib_38"]) / df["range"]
    df["dist_fib_50"] = (df["Close"] - df["fib_50"]) / df["range"]
    df["dist_fib_61"] = (df["Close"] - df["fib_61"]) / df["range"]

    # Break indicator
    df["break_fib_61"] = (df["Close"] < df["fib_61"]).astype(int)

    return df
```

```python
df = add_fibonacci_features(df, window=60)

# Drop rows with NaN (caused by rolling window)
df = df.dropna().reset_index(drop=True)
```
